```python
df = pd.read_csv("train.csv")

```

```python
# Task 3: First 10 rows
print("First 10 rows:")
display(df.head(10))

# Task 4: Last 5 rows
print("Last 5 rows:")
display(df.tail(5))

# Task 5: Dataset dimensions
print("Dataset Shape:")
print(df.shape)

# Task 6: Dataset information
print("Dataset Information:")
df.info()

# Task 7: Summary statistics
print("Summary Statistics:")
display(df.describe())

print("Age Statistics:")
print("Mean:", df["Age"].mean())
print("Minimum:", df["Age"].min())
print("Maximum:", df["Age"].max())

print("\nFare Statistics:")
print("Mean:", df["Fare"].mean())
print("Minimum:", df["Fare"].min())
print("Maximum:", df["Fare"].max())

# Task 8: Missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Task 9: Fill missing Age values with median

age_median = df["Age"].median()
df["Age"] = df["Age"].fillna(age_median)

print("Age median:", age_median)
print("Missing Age values after filling:", df["Age"].isnull().sum())
```

```python
# Task 10: Filter surviving passengers

survived_passengers = df[df["Survived"] == 1]

print("Number of surviving passengers:", len(survived_passengers))
display(survived_passengers.head())
```

```python
# Task 11: Convert Fare column to NumPy array

fare_array = df["Fare"].to_numpy()

print(fare_array)
```

```python
# Task 12: Maximum and average Fare

max_fare = np.max(fare_array)
average_fare = np.mean(fare_array)

print("Maximum Fare:", max_fare)
print("Average Fare:", average_fare)
```

```python
# Task 13: Histogram of Age

plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=20, color="skyblue", edgecolor="black")

plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
# Task 14: Bar chart of Survival Count

survival_counts = df["Survived"].value_counts().sort_index()

plt.figure(figsize=(6, 5))

plt.bar(["Did Not Survive", "Survived"],
        survival_counts.values,
        color=["red", "green"])

plt.title("Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
# Task 15: Scatter plot of Age vs Fare

plt.figure(figsize=(8, 5))

plt.scatter(df["Age"], df["Fare"],
            alpha=0.5,
            color="blue")

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()

```
